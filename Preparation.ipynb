# Lotka-Volterra
Fixed points exist at (F=0,R=0) and (F=400,R=900) for this configuration of Lotka-Volterra. 
Fixed points found by setting time derivatives in ODE system = 0.
At (F=0,R=0), fixed point will be unstable as eigenvalue is real.
At (F=400,R=900), fixed point will be stable with oscillatory time period 2*pi/omega = 10.47. Angular frequency omega=0.6 as eigen value lamda=0.6i, determined from det(J-lamda*I) = 0, where J is Jacobian representing the system of ODE's.

Fixed Point    | Stability | Oscillatory Time Period
(F=0, R=0)     | Unstable  | -
(F=400, R=900) | Stable    | 10.47 s

# Lorenz
Fixed points at: 
(x1,x2,x3) = (0,0,0)   
(x1,x2,x3) = (7*sqrt(8/3),7*sqrt(8/3),49)
(x1,x2,x3) = (-7*sqrt(8/3),-7*sqrt(8/3),49)

For (0,0,0), eigenvalue is real -> unstable
For the other two eigenvalues, numerical solving of polynomials are required.
det(J-lamda*I) = 0 for (x1,x2,x3) = (7*sqrt(8/3),7*sqrt(8/3),49) leads to 
-7848/3-160*lamda-41/3*lamda^2-lamda^3 = 0.
=> Im(lamda) => omega = 18.27
T = 2*pi/omega = 0.344 
same process for other fixed point

Fixed Point (x1,x2,x3)           | Stability | Oscillatory Time Period
(0, 0, 0)                        | Unstable  | -
(7*sqrt(8/3), 7*sqrt(8/3), 49)   | Stable    | 0.344 s
(-7*sqrt(8/3), -7*sqrt(8/3), 49) | Stable    | 0.590 s

```python
import numpy as np

# Solve Lorenz third degree polynomials numerically
lamda1 = np.roots([-1,-41/3,-160,-7848])
lamda2 = np.roots([-1,-41/3,-160,0])
print(f"Eigenvalues for fixed point (7*sqrt(8/3),7*sqrt(8/3),49): {lamda1}")
print(f"Eigenvalues for fixed point (-7*sqrt(8/3),-7*sqrt(8/3),49): {lamda2}")

```
