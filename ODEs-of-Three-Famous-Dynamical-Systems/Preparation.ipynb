# Lotka-Volterra
Fixed points exist at (F=0,R=0) and (F=400,R=900) for this configuration of Lotka-Volterra. 
Fixed points found by setting time derivatives in ODE system = 0.
At (F=0,R=0), fixed point will be unstable as eigenvalue is real.
At (F=400,R=900), fixed point will be stable with oscillatory time period 2*pi/omega = 10.47. Angular frequency omega=0.6 as eigen value lamda=0.6i, determined from det(J-lamda*I) = 0, where J is Jacobian representing the system of ODE's.

Fixed Point    | Stability | Oscillatory Time Period
(F=0, R=0)     | Unstable  | -
(F=400, R=900) | Stable    | 10.47 s

# Lorenz
Fixed points at: 
(x1,x2,x3) = (0,0,0)   
(x1,x2,x3) = (7*sqrt(8/3),7*sqrt(8/3),49)
(x1,x2,x3) = (-7*sqrt(8/3),-7*sqrt(8/3),49)

For (0,0,0), eigenvalue is real -> unstable
For the other two eigenvalues, numerical solving of polynomials are required.
det(J-lamda*I) = 0 for (x1,x2,x3) = (7*sqrt(8/3),7*sqrt(8/3),49) leads to 
-7848/3-160*lamda-41/3*lamda^2-lamda^3 = 0.
=> Im(lamda) => omega = 18.27
T = 2*pi/omega = 0.344 
same process for other fixed point

Fixed Point (x1,x2,x3)           | Stability | Oscillatory Time Period
(0, 0, 0)                        | Unstable  | -
(7*sqrt(8/3), 7*sqrt(8/3), 49)   | Stable    | 0.344 s
(-7*sqrt(8/3), -7*sqrt(8/3), 49) | Stable    | 0.590 s

```python
import numpy as np

# Solve Lorenz third degree polynomials numerically
beta = 8/3
rho = 50
sigma = 10
x1_fp2 = 7*np.sqrt(8/3)
x2_fp2 = 7*np.sqrt(8/3)
x3_fp2 = 49

lamda_fp2 = np.roots([-1, -sigma-1-beta, -sigma-sigma*beta-beta-x1_fp2**2+sigma*rho-sigma*x3_fp2, -sigma*beta-sigma*x1_fp2**2+sigma*rho*beta-sigma*x3_fp2*beta-sigma*x1_fp2*x2_fp2])

x1_fp3 = -7*np.sqrt(8/3)
x2_fp3 = -7*np.sqrt(8/3)
x3_fp3 = 49

lamda_fp3 = np.roots([-1, -sigma-1-beta, -sigma-sigma*beta-beta-x1_fp3**2+sigma*rho-sigma*x3_fp3, -sigma*beta-sigma*x1_fp3**2+sigma*rho*beta-sigma*x3_fp3*beta-sigma*x1_fp3*x2_fp3])

print(f"Eigenvalues for fixed point (7*sqrt(8/3),7*sqrt(8/3),49): {lamda_fp2}")
print(f"Eigenvalues for fixed point (-7*sqrt(8/3),-7*sqrt(8/3),49): {lamda_fp3}")

T = 2*np.pi/13.279448
print(T)
```

# Lotka-Volterra explanation
dR/dt = aR-bFR
dF/dt = cRF-dF

dR/dt: Net rate of growth for rabbits.
dF/dt: Net rate of growth for foxes.
aR: Rate at which rabbits reproduce. Depends on # of existing rabbits and a constant "a" which can be thought of as "How fast one rabbit becomes two rabbits on average"
bFR: Rate at which rabbits are killed by foxes. Depends on # of existing rabbits and # of existing foxes. Constant "b": "How fast 1 fox kills 1 rabbit on average."
cRF: The rate at which foxes reproduce. Depends on # of existing foxes and can thus reproduce. Also depends on # of existing rabbits which supply sustenance supporting reproduction. Constant "c": "How fast 1 fox becomes 2 foxes on average, given there exists one rabbit for food."
dF: The rate at which foxes die. Depends on # of existing foxes & constant "d": "How fast one fox dies on average".

b=c would essentially mean one fox eating one rabbit => one new fox produced. Which would likely not accurately represent reality. As a rabbit is smaller than a fox, this can not make sense in the long run. One would rather expect b>c, or "the rate at which 1 fox kills and eats 1 rabbit is greater than the rate at which foxes reproduce on average given one rabbit."
